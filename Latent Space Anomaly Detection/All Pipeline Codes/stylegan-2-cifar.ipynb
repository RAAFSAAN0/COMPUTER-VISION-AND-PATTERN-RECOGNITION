```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from torchvision import transforms, datasets
from torchvision.utils import save_image
import os
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.metrics import roc_auc_score, accuracy_score, confusion_matrix, ConfusionMatrixDisplay, roc_curve
from scipy import linalg
from scipy.stats import ttest_ind, mannwhitneyu
from sklearn.decomposition import PCA
from PIL import Image
import glob
from transformers import CLIPProcessor, CLIPModel
from sklearn.metrics import roc_auc_score, accuracy_score, confusion_matrix, ConfusionMatrixDisplay, roc_curve, auc



# === XAI RESULTS FOLDER ===
import os, csv, numpy as np, torch, matplotlib.pyplot as plt

# ensure folder
XAI_ROOT = "results/XAI"
os.makedirs(XAI_ROOT, exist_ok=True)
os.makedirs(f"{XAI_ROOT}/scores", exist_ok=True)

# Clear GPU memory before starting
torch.cuda.empty_cache()

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Hyperparameters and Dataset Choice
class Config:
    latent_dim = 512
    batch_size = 256
    lr = 0.0002
    epochs = 10
    xai_epochs = 10      
    baseline_epochs = 10
    epsilon_values = [0.1,0.3,0.5,0.7,0.9]
    attack_types = ['fgsm', 'pgd', 'cw', 'adaptive']
    detector_threshold = 0.5
    runs_for_stats = 5
    dataset = 'cifar10'
    n_critic = 5
    lambda_gp = 10
    weight_decay = 1e-5
    consistency_lambda = 0.1

config = Config()
img_channels = 3
img_size = 32
img_dim = img_channels * img_size * img_size


class EarlyStopping:
    def __init__(self, patience=3):
        self.patience = patience
        self.counter = 0
        self.best_loss = float('inf')
        self.should_stop = False

    def step(self, val_loss):
        if val_loss < self.best_loss:
            self.best_loss = val_loss
            self.counter = 0
        else:
            self.counter += 1
            if self.counter >= self.patience:
                self.should_stop = True


# Model Architectures

class MappingNetwork(nn.Module):
    def __init__(self, latent_dim, dlatent_size=512, num_layers=8):
        super().__init__()
        layers = []
        for _ in range(num_layers):
            layers.append(nn.Linear(latent_dim if _ == 0 else dlatent_size, dlatent_size))
            layers.append(nn.LeakyReLU(0.2))
        self.mapping = nn.Sequential(*layers)

    def forward(self, z):
        return self.mapping(z)

class StyleBlock(nn.Module):
    def __init__(self, in_channels, out_channels, style_dim=512):
        super().__init__()
        self.conv = nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1)
        self.style_fc = nn.Linear(style_dim, out_channels)  # ✅ inject style
        self.noise_weight = nn.Parameter(torch.zeros(1))
        self.bias = nn.Parameter(torch.zeros(out_channels))
        self.activation = nn.LeakyReLU(0.2)

    def forward(self, x, style):
        batch_size, _, h, w = x.shape
        noise = torch.randn(batch_size, 1, h, w, device=x.device)
        style_mod = self.style_fc(style).view(batch_size, -1, 1, 1)
        x = self.conv(x) + style_mod + self.noise_weight * noise + self.bias.view(1, -1, 1, 1)
        return self.activation(x)


class StyleGAN2Generator(nn.Module):
    def __init__(self, latent_dim=512):
        super().__init__()
        self.mapping = MappingNetwork(latent_dim)
        self.initial_const = nn.Parameter(torch.randn(1, 512, 4, 4))
        
        self.style1 = StyleBlock(512, 256)
        self.upsample1 = nn.Upsample(scale_factor=2)  

        self.style2 = StyleBlock(256, 128)
        self.upsample2 = nn.Upsample(scale_factor=2)  

        self.style3 = StyleBlock(128, 64)
        self.upsample3 = nn.Upsample(scale_factor=2) 

        self.to_rgb = nn.Conv2d(64, 3, kernel_size=1)

    def forward(self, z):
        style = self.mapping(z)
        x = self.initial_const.repeat(z.size(0), 1, 1, 1)
        x = self.style1(x, style)
        x = self.upsample1(x)
        x = self.style2(x, style)
        x = self.upsample2(x)
        x = self.style3(x, style)
        x = self.upsample3(x)
        x = self.to_rgb(x)
        return torch.tanh(x)


class StyleGAN2Discriminator(nn.Module):
    def __init__(self):
        super().__init__()

        def block(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, stride=2, padding=1),
                nn.LeakyReLU(0.2)
            )

        self.main = nn.Sequential(
            block(3, 64),     
            block(64, 128),    
            block(128, 256),   
            block(256, 512),  
            block(512, 512),   
        )
        self.flatten = nn.Flatten()
        self.fc = nn.Linear(512 * 1 * 1, 1)

    def forward(self, x):
        x = self.main(x)
        x = self.flatten(x)
        return self.fc(x)






class EnhancedXAIDetector(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()
        self.latent_dim = latent_dim
        self.feature_extractors = nn.ModuleList([
            self._build_feature_pathway(latent_dim, 64) for _ in range(4)
        ])
        self.cross_attention = nn.MultiheadAttention(
            embed_dim=64, num_heads=8, dropout=0.1
        )
        self.residual_blocks = nn.ModuleList([
            nn.Sequential(
                nn.Linear(64, 64),
                nn.LayerNorm(64),
                nn.ReLU(),
                nn.Dropout(0.1)
            ) for _ in range(3)
        ])
        self.classifier = nn.Sequential(
            nn.Linear(256, 128),
            nn.LayerNorm(128),
            nn.ReLU(),
            nn.Dropout(0.6),
            nn.Linear(128, 64),
            nn.LayerNorm(64),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
    
    def _build_feature_pathway(self, input_dim, output_dim):
        return nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.LayerNorm(256),
            nn.GELU(),
            nn.Dropout(0.2),
            nn.Linear(256, 128),
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Dropout(0.2),
            nn.Linear(128, output_dim)
        )
    
    def forward(self, z):
        features = []
        for extractor in self.feature_extractors:
            feat = extractor(z)
            features.append(feat)
        stacked_features = torch.stack(features, dim=1)
        attended_features, _ = self.cross_attention(
            stacked_features, stacked_features, stacked_features
        )
        for residual_block in self.residual_blocks:
            attended_features = attended_features + residual_block(attended_features)
        flattened = attended_features.view(attended_features.size(0), -1)
        return self.classifier(flattened)

class BaselineDetector(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()
        self.stat_extractor = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 32)
        )
        self.grad_extractor = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 32)
        )
        self.classifier = nn.Sequential(
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )
    
    def forward(self, z):
        stat_features = self.stat_extractor(z)
        grad_features = self.grad_extractor(z)
        combined = torch.cat([stat_features, grad_features], dim=1)
        return self.classifier(combined)

class MagNetDetector(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Conv2d(img_channels, 8, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(8, 16, kernel_size=3, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(16, 32, kernel_size=3, stride=2, padding=1),
            nn.ReLU()
        )
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(32, 16, kernel_size=4, stride=2, padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(16, 8, kernel_size=4, stride=2, padding=1),
            nn.ReLU(),
            nn.ConvTranspose2d(8, img_channels, kernel_size=4, stride=2, padding=1),
            nn.Tanh()
        )
        
    def forward(self, img):
        z = self.encoder(img)
        recon = self.decoder(z)
        return recon

    def reconstruction_error(self, img):
        recon = self.forward(img)
        error = ((img - recon)**2).mean(dim=[1,2,3])
        return error

class LatentExplainer(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 64),
            nn.ReLU(),
            nn.Linear(64, 8)  # Bottleneck
        )
        self.decoder = nn.Sequential(
            nn.Linear(8, 64),
            nn.ReLU(),
            nn.Linear(64, 256),
            nn.ReLU(),
            nn.Linear(256, latent_dim)
        )

    def forward(self, z):
        encoded = self.encoder(z)
        decoded = self.decoder(encoded)
       
        return decoded


class LatentVAE(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()
        self.fc1 = nn.Linear(latent_dim, 32)
        self.fc_mu = nn.Linear(32, 16)
        self.fc_logvar = nn.Linear(32, 16)
        self.fc3 = nn.Linear(16, 32)
        self.fc4 = nn.Linear(32, latent_dim)
        
    def encode(self, x):
        h1 = F.relu(self.fc1(x))
        return self.fc_mu(h1), self.fc_logvar(h1)
    
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    
    def decode(self, z):
        h3 = F.relu(self.fc3(z))
        return torch.tanh(self.fc4(h3))
    
    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        return self.decode(z), mu, logvar

class WGANAttacker:
    def __init__(self, G, C, detector=None):
        self.G = G
        self.C = C
        self.detector = detector

    def fgsm(self, z, epsilon=1):
        z = z.clone().detach().requires_grad_(True)
    
        self.G.train()
        self.C.train()
    
        gen_img = self.G(z)
        loss = -self.C(gen_img).mean()
    
        loss.backward()
    
        if z.grad is None:
            raise RuntimeError("FGSM failed: z.grad is None. Gradient did not flow.")
    
        return (z + epsilon * z.grad.sign()).detach()




    def pgd(self, z, epsilon=1, alpha=0.05, iterations=40):
        adv_z = z.clone().detach()
        for _ in range(iterations):
            adv_z = adv_z.clone().detach().requires_grad_(True)
            adv_z.retain_grad()  
            gen_img = self.G(adv_z)
            c_out = self.C(gen_img)
            loss = -c_out.mean()
            loss.backward()
            with torch.no_grad():
                adv_z = adv_z + alpha * adv_z.grad.sign()
                adv_z = torch.clamp(adv_z, z - epsilon, z + epsilon)
        return adv_z.detach()


    def cw(self, z, iterations=100, epsilon=1):
        """Carlini-Wagner L2-style attack – stronger version"""
        confidence = epsilon * 200  
        z_adv = z.clone().detach().requires_grad_(True)
        optimizer = optim.Adam([z_adv], lr=0.05)
        for _ in range(iterations):
            gen_img = self.G(z_adv)
            c_out = self.C(gen_img)
            loss = -c_out + confidence * torch.norm(z_adv - z, p=2)
            loss = loss.mean()
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
        return z_adv.detach()

    def enhanced_adaptive_attack(self, z, detector, epsilon=1, steps=60):
        """Enhanced Adaptive Attack – stronger version"""
        z_adv = z.clone().detach()
        detector_training = detector.training
        detector.eval()

        for _ in range(steps):
            z_adv = z_adv.clone().detach().requires_grad_(True)
            gen_img = self.G(z_adv)
            c_out = self.C(gen_img)
            det_out = detector(z_adv)

            lambda_det = 10.0  
            loss = -c_out.mean() + lambda_det * nn.BCELoss()(det_out, torch.zeros_like(det_out))

            loss.backward()
            alpha = epsilon * 0.25
            with torch.no_grad():
                z_adv = z_adv + alpha * z_adv.grad.sign()
                z_adv = torch.clamp(z_adv, z - epsilon, z + epsilon)

        if detector_training:
            detector.train()
        return z_adv.detach()


# Data Loading
class CIFAR10Dataset(torch.utils.data.Dataset):
    def __init__(self, root_dir='./data', transform=None, train=True, max_images=50000):
        self.dataset = datasets.CIFAR10(root=root_dir, train=train, download=True, transform=transform)
        self.max_images = min(max_images, len(self.dataset))

    def __len__(self):
        return self.max_images

    def __getitem__(self, idx):
        image, _ = self.dataset[idx]
        return image

transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
])
dataset = CIFAR10Dataset(root_dir='./data', transform=transform, train=True, max_images=20000)
train_loader = DataLoader(
    dataset,
    batch_size=config.batch_size,
    shuffle=True,
    num_workers=8,
    pin_memory=True,
    persistent_workers=True
)

from torch.utils.data import random_split

magnet_train_size = int(0.8 * len(train_loader.dataset))
magnet_val_size = len(train_loader.dataset) - magnet_train_size
magnet_train_set, magnet_val_set = random_split(train_loader.dataset, [magnet_train_size, magnet_val_size])

magnet_train_loader = DataLoader(magnet_train_set, batch_size=32, shuffle=True)
magnet_val_loader = DataLoader(magnet_val_set, batch_size=32, shuffle=False)


# Gradient Penalty for WGAN-GP
def gradient_penalty(critic, real, fake, device):
    batch_size, C, H, W = real.shape
    alpha = torch.rand(batch_size, 1, 1, 1, device=device).expand_as(real)
    interpolates = alpha * real + (1 - alpha) * fake
    interpolates = interpolates.requires_grad_(True)
    critic_inter = critic(interpolates)
    gradients = torch.autograd.grad(outputs=critic_inter, inputs=interpolates,
                                    grad_outputs=torch.ones_like(critic_inter),
                                    create_graph=True, retain_graph=True)[0]
    gradients = gradients.view(batch_size, -1)
    gp = ((gradients.norm(2, dim=1) - 1) ** 2).mean()
    return gp

# WGAN-GP Training
def train_stylegan2():
    G.train()
    C.train()
    
    optimizer_G = optim.Adam(G.parameters(), lr=config.lr, betas=(0.0, 0.99))
    optimizer_D = optim.Adam(C.parameters(), lr=config.lr, betas=(0.0, 0.99))

    pbar = tqdm(range(config.epochs), desc="Training StyleGAN2")
    for epoch in pbar:
        d_loss_epoch = 0.0
        g_loss_epoch = 0.0

        for batch_idx, real_imgs in enumerate(train_loader):
            real_imgs = real_imgs.to(device)
            batch_size = real_imgs.size(0)

            # === Train Discriminator (n_critic times) ===
            for _ in range(config.n_critic):
                z = torch.randn(batch_size, config.latent_dim).to(device)
                fake_imgs = G(z).detach()

                real_scores = C(real_imgs)
                fake_scores = C(fake_imgs)

                gp = gradient_penalty(C, real_imgs, fake_imgs, device)

                d_loss = -torch.mean(real_scores) + torch.mean(fake_scores) + config.lambda_gp * gp

                optimizer_D.zero_grad()
                d_loss.backward()
                optimizer_D.step()
                d_loss_epoch += d_loss.item()

            # === Train Generator ===
            z = torch.randn(batch_size, config.latent_dim).to(device)
            fake_imgs = G(z)
            fake_scores = C(fake_imgs)

            g_loss = -torch.mean(fake_scores)

            optimizer_G.zero_grad()
            g_loss.backward()
            optimizer_G.step()
            g_loss_epoch += g_loss.item()

        avg_d_loss = d_loss_epoch / len(train_loader)
        avg_g_loss = g_loss_epoch / len(train_loader)

        pbar.set_postfix({'G Loss': avg_g_loss, 'D Loss': avg_d_loss})

        # Save generated samples every 10 epochs
        if (epoch + 1) % 10 == 0:
            with torch.no_grad():
                z_sample = torch.randn(64, config.latent_dim).to(device)
                gen_samples = G(z_sample)
                os.makedirs("results", exist_ok=True)
                save_image((gen_samples + 1) / 2, f"results/gen_epoch_{epoch+1}.png", nrow=8)

    torch.cuda.empty_cache()

     
    


    # # ✅ Ensure the results directory exists
    # os.makedirs("results", exist_ok=True)

    # # Plot and save loss curves
    # plt.figure(figsize=(10, 6))
    # plt.plot(C_losses, label="Critic Loss", color='r')
    # plt.plot(G_losses, label="Generator Loss", color='b')
    # plt.xlabel('Epochs')
    # plt.ylabel('Loss')
    # plt.title('WGAN-GP Loss Curve')
    # plt.legend()
    # plt.savefig(os.path.join('results', 'wgan_loss_curve.png'))
    # plt.show()


# Explanation Consistency Regularization
def compute_explanation_consistency(detector, clean_z, adv_z):
    detector.eval()
    with torch.no_grad():
        clean_pred = detector(clean_z)
        adv_pred = detector(adv_z)
    consistency_loss = F.mse_loss(clean_pred, adv_pred)
    return consistency_loss

# Detector Data Generation
def generate_detector_data(attack_type='all', epsilon=0.1):
    normal_z = torch.randn(500, config.latent_dim, device=device)
    adversarial_z = []

    G.train()
    C.train()

    if attack_type == 'all':
        attacks = {
            'fgsm': lambda z: attacker.fgsm(z),
            'pgd': lambda z: attacker.pgd(z.clone().detach(), epsilon=epsilon),
            'cw': lambda z: attacker.cw(z.clone().detach(), epsilon=epsilon),
            'adaptive': lambda z: attacker.enhanced_adaptive_attack(z.clone().detach(), detector, epsilon=epsilon)
        }
    else:
        fn = getattr(attacker, attack_type)
        if attack_type == 'adaptive':
            attacks = {attack_type: lambda z: fn(z.clone().detach(), detector, epsilon=epsilon)}
        else:
            attacks = {attack_type: lambda z: fn(z)}

    for attack_name, attack_fn in attacks.items():
        adv_chunks = []
        for chunk in normal_z.chunk(10):
            chunk = chunk.clone().detach().to(device)
            adv = attack_fn(chunk)
            adv_chunks.append(adv)
        adversarial_z.append(torch.cat(adv_chunks))

    X = torch.cat([normal_z] + adversarial_z)
    y = torch.cat([
        torch.zeros(len(normal_z)),
        *[torch.ones(len(adv)) for adv in adversarial_z]
    ]).view(-1, 1).to(device)

    return X, y




# Robust XAI Detector Training
def train_robust_xai_detector(detector, generator, epochs=config.xai_epochs):
    optimizer = optim.AdamW(detector.parameters(), lr=5e-5, weight_decay=config.weight_decay)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, epochs)
    criterion = nn.BCELoss()
    early_stopper = EarlyStopping(patience=3)

    pbar = tqdm(range(epochs), desc="Training EnhancedXAIDetector")
    for epoch in pbar:
        detector.train()
        epoch_loss = 0
        val_loss = 0

        for batch_idx, real_imgs in enumerate(train_loader):
            batch_size = real_imgs.size(0)
            clean_z = torch.randn(batch_size, config.latent_dim).to(device)

            # Label smoothing
            eps = 0.1
            adv_z_fgsm = attacker.fgsm(clean_z, epsilon=min(0.1 + 0.01 * epoch, 0.3))
            adv_z_pgd = attacker.pgd(clean_z, epsilon=min(0.1 + 0.01 * epoch, 0.3))
            adv_z_adaptive = attacker.enhanced_adaptive_attack(clean_z, detector, epsilon=min(0.1 + 0.01 * epoch, 0.3))

            all_z = torch.cat([clean_z, adv_z_fgsm, adv_z_pgd, adv_z_adaptive])
            all_labels = torch.cat([
                torch.full((len(clean_z), 1), eps),
                torch.full((len(adv_z_fgsm), 1), 1 - eps),
                torch.full((len(adv_z_pgd), 1), 1 - eps),
                torch.full((len(adv_z_adaptive), 1), 1 - eps)
            ]).to(device)

            predictions = detector(all_z)
            detection_loss = criterion(predictions, all_labels)

            if batch_idx % 5 == 0:
                consistency_loss = compute_explanation_consistency(detector, clean_z, adv_z_adaptive)
                total_loss = detection_loss + config.consistency_lambda * consistency_loss
                val_loss += total_loss.item()
            else:
                total_loss = detection_loss

            optimizer.zero_grad()
            total_loss.backward()
            torch.nn.utils.clip_grad_norm_(detector.parameters(), 1.0)
            optimizer.step()
            epoch_loss += total_loss.item()

        avg_loss = epoch_loss / len(train_loader)
        avg_val_loss = val_loss / (len(train_loader) // 5)
        scheduler.step()

        pbar.set_postfix({'Train Loss': avg_loss, 'Val Loss': avg_val_loss})
        early_stopper.step(avg_val_loss)
        if early_stopper.should_stop:
            print(f"[Early Stopping] XAI Detector stopped at epoch {epoch+1}")
            break
    torch.cuda.empty_cache()


from sklearn.model_selection import train_test_split

def train_baseline_detector(attack_type='all', epsilon=0.1, epochs=config.baseline_epochs):
    # ✅ Generate latent vectors and labels
    X, y = generate_detector_data(attack_type, epsilon)

    # ✅ Perform train/val split
    X_train, X_val, y_train, y_val = train_test_split(X.cpu(), y.cpu(), test_size=0.2, random_state=42)
    X_train, X_val, y_train, y_val = X_train.to(device), X_val.to(device), y_train.to(device), y_val.to(device)

    # ✅ Create data loaders
    train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=32, shuffle=True)
    val_loader = DataLoader(TensorDataset(X_val, y_val), batch_size=32)

    # ✅ Optimizer and loss
    optimizer = optim.AdamW(baseline_detector.parameters(), lr=5e-5, weight_decay=config.weight_decay)
    criterion = nn.BCELoss()
    early_stopper = EarlyStopping(patience=3)

    pbar = tqdm(range(epochs), desc="Training BaselineDetector")
    for epoch in pbar:
        baseline_detector.train()
        epoch_loss = 0
        for batch_X, batch_y in train_loader:
            batch_X, batch_y = batch_X.to(device), batch_y.to(device)
            outputs = baseline_detector(batch_X)
            loss = criterion(outputs, batch_y)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()

        # ✅ Validation
        val_loss = 0
        baseline_detector.eval()
        with torch.no_grad():
            for val_X, val_y in val_loader:
                val_X, val_y = val_X.to(device), val_y.to(device)
                val_outputs = baseline_detector(val_X)
                val_loss += criterion(val_outputs, val_y).item()
        val_loss /= len(val_loader)

        pbar.set_postfix({'Train Loss': epoch_loss / len(train_loader), 'Val Loss': val_loss})
        early_stopper.step(val_loss)
        if early_stopper.should_stop:
            print(f"[Early Stopping] Baseline Detector stopped at epoch {epoch+1}")
            break

    torch.cuda.empty_cache()

     

def train_latent_explainer(explainer, attacker, generator, detector=None, epochs=20, epsilon=0.1, lambda_consistency=0.1):
    explainer.train()
    optimizer = optim.AdamW(explainer.parameters(), lr=1e-3, weight_decay=config.weight_decay)
    criterion = nn.L1Loss()

    dataset = torch.randn(5000, config.latent_dim).to(device)
    loader = DataLoader(TensorDataset(dataset), batch_size=32, shuffle=True)

    for epoch in range(epochs):
        total_loss = 0
        for batch in loader:
            z_clean = batch[0].to(device)
            z_adv = attacker.enhanced_adaptive_attack(z_clean, detector, epsilon=epsilon)

            all_z = torch.cat([z_clean, z_adv], dim=0)
            recon_all = explainer(all_z)

            recon_loss = criterion(recon_all, all_z)

            recon_clean = explainer(z_clean)
            recon_adv = explainer(z_adv)
            consistency_loss = F.mse_loss(recon_clean, z_clean) - F.mse_loss(recon_adv, z_adv)

            loss = recon_loss - lambda_consistency * consistency_loss

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            total_loss += loss.item()
        print(f"[LatentExplainer Epoch {epoch+1}] Loss: {total_loss / len(loader):.4f}")

    torch.cuda.empty_cache()




def train_latent_vae(vae, epochs=1):
    vae.train()
    optimizer = optim.AdamW(vae.parameters(), lr=1e-3, weight_decay=config.weight_decay)
    def vae_loss(recon_x, x, mu, logvar):
        recon_loss = F.mse_loss(recon_x, x, reduction='mean')
        kl_loss = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
        return recon_loss + kl_loss
    dataset = torch.randn(5000, config.latent_dim).to(device)
    loader = DataLoader(TensorDataset(dataset), batch_size=32, shuffle=True)
    for epoch in range(epochs):
        total_loss = 0
        for batch in loader:
            x = batch[0]
            recon_x, mu, logvar = vae(x)
            loss = vae_loss(recon_x, x, mu, logvar)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        print(f"[LatentVAE Epoch {epoch+1}] Loss: {total_loss/len(loader):.4f}")
    torch.cuda.empty_cache()


def train_magnet_detector(num_epochs=10):
    optimizer = optim.AdamW(magnet_detector.parameters(), lr=0.001, weight_decay=config.weight_decay)
    criterion = nn.MSELoss()

    pbar = tqdm(range(num_epochs), desc="Training MagNet")
    for epoch in pbar:
        # === Training ===
        magnet_detector.train()
        train_loss = 0
        for imgs in magnet_train_loader:  # ✅ correct: no label to unpack
            imgs = imgs.to(device)
            recon = magnet_detector(imgs)
            loss = criterion(recon, imgs)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        train_loss /= len(magnet_train_loader)

        # === Validation ===
        magnet_detector.eval()
        val_loss = 0
        with torch.no_grad():
            for imgs in magnet_val_loader:  # ✅ correct here too
                imgs = imgs.to(device)
                recon = magnet_detector(imgs)
                loss = criterion(recon, imgs)
                val_loss += loss.item()
        val_loss /= len(magnet_val_loader)

        pbar.set_postfix({'Train Loss': train_loss, 'Val Loss': val_loss})

    torch.cuda.empty_cache()




    
from sklearn.metrics import roc_curve, auc, accuracy_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay

def evaluate_baseline_detector(attack_type='all', epsilon=0.1, save_dir="results"):
    os.makedirs(save_dir, exist_ok=True)

    test_z = torch.randn(200, config.latent_dim).to(device)

    if attack_type == 'all':
        attacks = [attacker.fgsm, attacker.pgd, attacker.cw, lambda z: attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon)]
    else:
        attacks = [lambda z: attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon) if attack_type == 'adaptive' else getattr(attacker, attack_type)]

    all_z = [test_z]
    labels = [torch.zeros(len(test_z))]
    for attack_fn in attacks:
        adv_z = torch.cat([attack_fn(z) for z in test_z.chunk(10)])
        all_z.append(adv_z)
        labels.append(torch.ones(len(adv_z)))

    X_test = torch.cat(all_z)
    y_test = torch.cat(labels).cpu().numpy()

    with torch.no_grad():
        scores = baseline_detector(X_test).cpu().numpy().flatten()

    # Optimal threshold from ROC
    fpr, tpr, thresholds = roc_curve(y_test, scores)
    best_thresh = thresholds[np.argmax(tpr - fpr)]
    print(f"[INFO] Baseline best threshold from ROC: {best_thresh:.4f}")

    y_pred = (scores > best_thresh).astype(int)
    auroc = roc_auc_score(y_test, scores)
    acc = accuracy_score(y_test, y_pred)

    # Plot ROC + CM
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))
    ax[0].plot(fpr, tpr, label=f'AUC = {auroc:.2f}')
    ax[0].plot([0, 1], [0, 1], 'k--')
    ax[0].set_xlabel("FPR")
    ax[0].set_ylabel("TPR")
    ax[0].set_title("BaselineDetector ROC Curve")
    ax[0].legend()

    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(cm, display_labels=['Normal', 'Adv'])
    disp.plot(ax=ax[1])
    ax[1].set_title("BaselineDetector Confusion Matrix")

    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, f"baseline_evaluation_plots_eps_{epsilon}.png"))
    plt.show()

    print(f"Baseline Detector: AUROC: {auroc:.4f}, Accuracy: {acc:.4f}")
    return auroc, acc

  

def evaluate_latent_explainer_roc_all_eps(explainer, save_dir="results"):
    os.makedirs(save_dir, exist_ok=True)
    aurocs = []
    for epsilon in config.epsilon_values:
        normal_z = torch.randn(200, config.latent_dim).to(device)
        adv_z = torch.cat([
            attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon)
            for z in normal_z.chunk(10)
        ])
        with torch.no_grad():
            recon_normal = explainer(normal_z)
            recon_adv = explainer(adv_z)
        err_normal = F.mse_loss(recon_normal, normal_z, reduction='none').mean(dim=1).cpu().numpy()
        err_adv = F.mse_loss(recon_adv, adv_z, reduction='none').mean(dim=1).cpu().numpy()
        scores = np.concatenate([err_normal, err_adv])
        labels = np.concatenate([np.zeros(len(err_normal)), np.ones(len(err_adv))])
        fpr, tpr, _ = roc_curve(labels, scores)
        auroc = roc_auc_score(labels, scores)
        aurocs.append((epsilon, auroc))

        # Plot ROC curve
        plt.figure(figsize=(8,6))
        plt.plot(fpr, tpr, label=f'AUROC = {auroc:.2f}')
        plt.plot([0,1], [0,1], 'k--')
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title(f"ROC Curve - LatentExplainer (ε={epsilon})")
        plt.legend()
        plt.tight_layout()
        plt.savefig(os.path.join(save_dir, f"latent_explainer_roc_eps_{epsilon}.png"))
        plt.close()

        print(f"[LatentExplainer] ε={epsilon} -> AUROC: {auroc:.4f}")
    return aurocs
  
  

  
def evaluate_latent_vae_roc_all_eps(vae, save_dir="results"):
    os.makedirs(save_dir, exist_ok=True)
    aurocs = []
    for epsilon in config.epsilon_values:
        normal_z = torch.randn(200, config.latent_dim).to(device)
        adv_z = torch.cat([
            attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon)
            for z in normal_z.chunk(10)
        ])
        with torch.no_grad():
            recon_normal, _, _ = vae(normal_z)
            recon_adv, _, _ = vae(adv_z)
        err_normal = F.mse_loss(recon_normal, normal_z, reduction='none').mean(dim=1).cpu().numpy()
        err_adv = F.mse_loss(recon_adv, adv_z, reduction='none').mean(dim=1).cpu().numpy()
        scores = np.concatenate([err_normal, err_adv])
        labels = np.concatenate([np.zeros(len(err_normal)), np.ones(len(err_adv))])
        fpr, tpr, _ = roc_curve(labels, scores)
        auroc = roc_auc_score(labels, scores)
        aurocs.append((epsilon, auroc))

        # Plot ROC curve
        plt.figure(figsize=(8,6))
        plt.plot(fpr, tpr, label=f'AUROC = {auroc:.2f}')
        plt.plot([0,1], [0,1], 'k--')
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title(f"ROC Curve - LatentVAE (ε={epsilon})")
        plt.legend()
        plt.tight_layout()
        plt.savefig(os.path.join(save_dir, f"latent_vae_roc_eps_{epsilon}.png"))
        plt.close()

        print(f"[LatentVAE] ε={epsilon} -> AUROC: {auroc:.4f}")
    return aurocs

def plot_auroc_vs_epsilon(xid_results, vae_results, save_path="results/auroc_vs_epsilon.png"):
    eps_xid, auroc_xid = zip(*sorted(xid_results))
    eps_vae, auroc_vae = zip(*sorted(vae_results))

    plt.figure(figsize=(8, 6))
    plt.plot(eps_xid, auroc_xid, marker='o', label='LatentExplainer', color='blue')
    plt.plot(eps_vae, auroc_vae, marker='s', label='LatentVAE', color='green')
    plt.xlabel("Epsilon (ε)")
    plt.ylabel("AUROC")
    plt.title("AUROC vs Epsilon")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    plt.savefig(save_path)
    plt.show()



def visualize_magnet_detector(magnet, save_dir="results"):
    magnet.eval()
    os.makedirs(save_dir, exist_ok=True)
    
    # 1. Original vs Reconstructed Images
    real_batch = next(iter(train_loader))
    real_imgs = real_batch[:8].to(device)
    with torch.no_grad():
        recon_imgs = magnet(real_imgs)
    real_imgs_np = real_imgs.cpu().numpy() * 0.5 + 0.5
    recon_imgs_np = recon_imgs.cpu().numpy() * 0.5 + 0.5
    fig, axs = plt.subplots(2, 8, figsize=(16, 4))
    for i in range(8):
        axs[0, i].imshow(real_imgs_np[i].transpose(1, 2, 0))
        axs[0, i].set_title("Original")
        axs[0, i].axis('off')
        axs[1, i].imshow(recon_imgs_np[i].transpose(1, 2, 0))
        axs[1, i].set_title("Reconstructed")
        axs[1, i].axis('off')
    plt.suptitle("MagNetDetector: Original vs Reconstructed Images")
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, "magnet_recon_images.png"))
    plt.show()
    
    # 2. Reconstruction Error Histogram
    normal_z = torch.randn(200, config.latent_dim).to(device)
    adv_z = torch.cat([attacker.enhanced_adaptive_attack(z, detector, epsilon=0.1) for z in normal_z.chunk(10)])
    normal_imgs = G(normal_z)
    adv_imgs = G(adv_z)
    with torch.no_grad():
        normal_errors = magnet.reconstruction_error(normal_imgs)
        adv_errors = magnet.reconstruction_error(adv_imgs)
    plt.figure(figsize=(8, 6))
    plt.hist(normal_errors.cpu().numpy(), bins=30, alpha=0.5, label="Normal", color='blue')
    plt.hist(adv_errors.cpu().numpy(), bins=30, alpha=0.5, label="Adversarial", color='red')
    plt.title("MagNetDetector: Reconstruction Error Distribution")
    plt.xlabel("Reconstruction Error")
    plt.ylabel("Frequency")
    plt.legend()
    plt.savefig(os.path.join(save_dir, "magnet_recon_error.png"))
    plt.show()
    
    # 3. ROC Curve for Anomaly Detection
    labels = np.hstack([np.zeros(len(normal_errors)), np.ones(len(adv_errors))])
    scores = np.hstack([normal_errors.cpu().numpy(), adv_errors.cpu().numpy()])
    fpr, tpr, _ = roc_curve(labels, scores)
    auroc = roc_auc_score(labels, scores)
    plt.figure(figsize=(8, 6))
    plt.plot(fpr, tpr, label=f'AUROC = {auroc:.2f}')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("MagNetDetector: ROC Curve")
    plt.legend()
    plt.savefig(os.path.join(save_dir, "magnet_roc_curve.png"))
    plt.show()

# Anomaly Detection
def detect_latent_anomalies(explainer, z, threshold=0.1, lambda_entropy=0.1):
    explainer.eval()
    with torch.no_grad():
        z_recon = explainer(z)
        deviation = torch.norm(z - z_recon, dim=1)
        entropy = -torch.sum(F.softmax(z, dim=1) * F.log_softmax(z, dim=1), dim=1)
        anomaly_score = deviation + lambda_entropy * entropy
        return anomaly_score, anomaly_score > threshold







def evaluate_magnet_detector_all_eps(magnet, G, save_dir="results"):
    magnet.eval()
    G.eval()
    os.makedirs(save_dir, exist_ok=True)

    for epsilon in config.epsilon_values:
        print(f"\n[MagNet] Evaluating for ε = {epsilon}")

        # Generate latent vectors
        z = torch.randn(200, config.latent_dim).to(device)
        x = G(z)
        adv_z = attacker.enhanced_adaptive_attack(z, detector, epsilon)
        x_adv = G(adv_z)

        with torch.no_grad():
            recon_x = magnet(x)
            recon_x_adv = magnet(x_adv)

        # Compute reconstruction error
        normal_scores = torch.mean((recon_x - x) ** 2, dim=[1, 2, 3]).detach().cpu().numpy()
        adv_scores = torch.mean((recon_x_adv - x_adv) ** 2, dim=[1, 2, 3]).detach().cpu().numpy()


        # Labels and scores
        all_scores = np.concatenate([normal_scores, adv_scores])
        labels = np.concatenate([np.zeros_like(normal_scores), np.ones_like(adv_scores)])

        # 🎯 ROC Curve
        fpr, tpr, _ = roc_curve(labels, all_scores)
        roc_auc = auc(fpr, tpr)

        plt.figure()
        plt.plot(fpr, tpr, label=f'AUROC = {roc_auc:.4f}')
        plt.plot([0, 1], [0, 1], 'k--')
        plt.title(f"MagNet ROC Curve (ε={epsilon})")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.legend(loc="lower right")
        roc_path = os.path.join(save_dir, f"magnet_roc_eps_{epsilon}.png")
        plt.savefig(roc_path)
        plt.show()
        plt.close()
        print(f"[✓] ROC saved: {roc_path}")

        # 📊 Histogram
        plt.figure()
        plt.hist(normal_scores, bins=30, alpha=0.6, label='Normal', color='blue')
        plt.hist(adv_scores, bins=30, alpha=0.6, label='Adversarial', color='red')
        plt.title(f"MagNet Reconstruction Error Histogram (ε={epsilon})")
        plt.xlabel("Reconstruction Error")
        plt.ylabel("Count")
        plt.legend()
        hist_path = os.path.join(save_dir, f"magnet_histogram_eps_{epsilon}.png")
        plt.savefig(hist_path)
        plt.show()
        plt.close()
        print(f"[✓] Histogram saved: {hist_path}")
def multi_modal_detection(explainer, z, G, C, alpha=0.5, threshold=0.1):
    explainer.eval()
    with torch.no_grad():
        z_recon = explainer(z)
        deviation = torch.norm(z - z_recon, dim=1)
        entropy = -torch.sum(F.softmax(z, dim=1) * F.log_softmax(z, dim=1), dim=1)
        anomaly_score = deviation + 0.01 * entropy
        gen_imgs = G(z)
        disc_confidence = C(gen_imgs).view(-1)
        combined_score = alpha * anomaly_score + (1-alpha) * disc_confidence.abs()
        return combined_score, combined_score > threshold

def adaptive_threshold(scores, base_threshold=0.5, epsilon=0.1):
    mean_score = scores.mean().item()
    adjusted_threshold = base_threshold + epsilon * mean_score
    return adjusted_threshold

# FID Calculation with CLIP
def compute_fid(real_images, gen_images):
    from torchvision.transforms import Resize
    resize = Resize((224, 224))
    model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
    processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
    model.eval()
    
    def get_features(x):
        with torch.no_grad():
            x = resize(x)
            x = x.cpu().numpy()
            x = np.transpose(x, (0, 2, 3, 1)) * 0.5 + 0.5
            x = [Image.fromarray((img * 255).astype(np.uint8)) for img in x]

            inputs = processor(images=x, return_tensors="pt", padding=True)
            inputs = {k: v.to(device) for k, v in inputs.items()}
            features = model.get_image_features(**inputs)
        return features
    
    real = torch.tensor(real_images).to(device)
    gen = torch.tensor(gen_images).to(device)
    act1 = get_features(real).cpu().numpy()
    act2 = get_features(gen).cpu().numpy()
    mu1, sigma1 = np.mean(act1, axis=0), np.cov(act1, rowvar=False)
    mu2, sigma2 = np.mean(act2, axis=0), np.cov(act2, rowvar=False)
    diff = mu1 - mu2
    covmean, _ = linalg.sqrtm(sigma1.dot(sigma2), disp=False)
    if np.iscomplexobj(covmean):
        covmean = covmean.real
    fid = diff.dot(diff) + np.trace(sigma1 + sigma2 - 2*covmean)
    return fid

from sklearn.metrics import roc_auc_score, accuracy_score, confusion_matrix, ConfusionMatrixDisplay, roc_curve

def evaluate_detector(attack_type='all', epsilon=0.1, adaptive=False, use_multi_modal=False, save_dir="results"):
    os.makedirs(save_dir, exist_ok=True)
    test_z = torch.randn(200, config.latent_dim).to(device)

    # ✅ Generate adversarial samples
    if attack_type == 'all':
        attacks = [attacker.fgsm, attacker.pgd, attacker.cw, lambda z: attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon)]
    else:
        attacks = [lambda z: attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon) if attack_type == 'adaptive' else getattr(attacker, attack_type)]

    all_z = [test_z]
    labels = [torch.zeros(len(test_z))]

    for attack_fn in attacks:
        adv_z = torch.cat([attack_fn(z) for z in test_z.chunk(10)])
        all_z.append(adv_z)
        labels.append(torch.ones(len(adv_z)))

    X_test = torch.cat(all_z)
    y_test = torch.cat(labels).cpu().numpy()

    # ✅ Score the samples
    if use_multi_modal:
        scores, _ = multi_modal_detection(latent_explainer, X_test, G, C, threshold=0.0)
        scores = scores.cpu().numpy().flatten()
    else:
        with torch.no_grad():
            scores = detector(X_test).cpu().numpy().flatten()

    # ✅ Find best threshold from ROC curve
    fpr, tpr, thresholds = roc_curve(y_test, scores)
    best_thresh = thresholds[np.argmax(tpr - fpr)]
    print(f"[INFO] Best threshold from ROC: {best_thresh:.4f}")

    y_pred = (scores > best_thresh).astype(int)
    auroc = roc_auc_score(y_test, scores)
    acc = accuracy_score(y_test, y_pred)

    # ✅ FID calculation
    with torch.no_grad():
        gen_imgs = G(test_z).cpu().numpy()
        real_batch = next(iter(train_loader))
        real_imgs = real_batch[:len(test_z)].cpu().numpy()
        gen_imgs = (gen_imgs + 1)/2.
        real_imgs = (real_imgs + 1)/2.
    fid_score = compute_fid(real_imgs, gen_imgs)

    # ✅ Plot ROC and Confusion Matrix
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))
    ax[0].plot(fpr, tpr, label=f'AUC = {auroc:.2f}')
    ax[0].plot([0, 1], [0, 1], 'k--')
    ax[0].set_xlabel('FPR')
    ax[0].set_ylabel('TPR')
    ax[0].set_title('ROC Curve')
    ax[0].legend()

    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(cm, display_labels=['Normal', 'Adv'])
    disp.plot(ax=ax[1])
    ax[1].set_title('Confusion Matrix')
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, f"evaluation_plots_eps_{epsilon}.png"))
    plt.show()

    # ✅ Print results
    print(f"Evaluation for {attack_type.upper()} attack (ε={epsilon}{' adaptive' if adaptive else ''}):")
    print(f"AUROC: {auroc:.4f}, Accuracy: {acc:.4f}, FID: {fid_score:.4f}")
    return auroc, acc, fid_score


def visualize_latent_space(save_dir="results"):
    os.makedirs(save_dir, exist_ok=True)
    
    normal_z = torch.randn(200, config.latent_dim).cpu().numpy()
    adv_z = []
    
    # Generate adversarial examples from multiple attacks
    for attack_fn in [attacker.fgsm, attacker.pgd, attacker.cw, lambda z: attacker.enhanced_adaptive_attack(z, detector, epsilon=0.1)]:
        adv_z_part = torch.cat([
            attack_fn(z) for z in torch.tensor(normal_z).to(device).chunk(10)
        ]).cpu().numpy()
        adv_z.append(adv_z_part)
        
    adv_z = np.concatenate(adv_z, axis=0)
    
    # Labels: 0 = normal, 1 = adversarial
    labels = np.hstack([np.zeros(len(normal_z)), np.ones(len(adv_z))])
    codes = np.vstack([normal_z, adv_z])
    
    # PCA projection
    from sklearn.decomposition import PCA
    pca = PCA(n_components=2)
    proj = pca.fit_transform(codes)
    
    # Plotting
    plt.figure(figsize=(8, 6))
    plt.scatter(proj[labels==0, 0], proj[labels==0, 1],
                label="Normal", alpha=0.8, color='blue', edgecolor='black', marker='o')
    plt.scatter(proj[labels==1, 0], proj[labels==1, 1],
                label="Adversarial", alpha=0.5, color='orange', edgecolor='black', marker='x')
    
    plt.title("Latent Space (PCA projection)")
    plt.xlabel("PCA Component 1")
    plt.ylabel("PCA Component 2")
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, "latent_space_pca.png"))
    plt.show()



def visualize_latent_explainer_histograms_all_eps(explainer, save_dir="results"):
    explainer.eval()
    os.makedirs(save_dir, exist_ok=True)

    for epsilon in config.epsilon_values:
        normal_z = torch.randn(200, config.latent_dim).to(device)
        adv_z = torch.cat([
            attacker.enhanced_adaptive_attack(z, detector, epsilon)  # positional arg
            for z in normal_z.chunk(10)
        ])
        with torch.no_grad():
            recon_normal = explainer(normal_z)
            recon_adv = explainer(adv_z)
        err_normal = F.mse_loss(recon_normal, normal_z, reduction='none').mean(dim=1).cpu().numpy()
        err_adv = F.mse_loss(recon_adv, adv_z, reduction='none').mean(dim=1).cpu().numpy()

        plt.figure(figsize=(8, 6))
        plt.hist(err_normal, bins=30, alpha=0.5, label="Normal", color='blue')
        plt.hist(err_adv, bins=30, alpha=0.5, label="Adversarial", color='red')
        plt.title(f"LatentExplainer: Reconstruction Error (ε={epsilon})")
        plt.xlabel("Reconstruction Error")
        plt.ylabel("Frequency")
        plt.legend()
        plt.tight_layout()
        
        save_path = os.path.join(save_dir, f"latent_explainer_recon_error_eps_{epsilon}.png")
        plt.savefig(save_path)
        plt.show()  # <-- Show the histogram on screen
        plt.close()
        print(f"[✓] Saved and displayed: {save_path}")


def visualize_latent_vae_histograms_all_eps(vae, save_dir="results"):
    vae.eval()
    os.makedirs(save_dir, exist_ok=True)

    for epsilon in config.epsilon_values:
        normal_z = torch.randn(200, config.latent_dim).to(device)
        adv_z = torch.cat([
            attacker.enhanced_adaptive_attack(z, detector, epsilon)  # positional arg
            for z in normal_z.chunk(10)
        ])
        with torch.no_grad():
            recon_normal, _, _ = vae(normal_z)
            recon_adv, _, _ = vae(adv_z)
        err_normal = F.mse_loss(recon_normal, normal_z, reduction='none').mean(dim=1).cpu().numpy()
        err_adv = F.mse_loss(recon_adv, adv_z, reduction='none').mean(dim=1).cpu().numpy()

        plt.figure(figsize=(8, 6))
        plt.hist(err_normal, bins=30, alpha=0.5, label="Normal", color='blue')
        plt.hist(err_adv, bins=30, alpha=0.5, label="Adversarial", color='red')
        plt.title(f"LatentVAE: Reconstruction Error (ε={epsilon})")
        plt.xlabel("Reconstruction Error")
        plt.ylabel("Frequency")
        plt.legend()
        plt.tight_layout()

        save_path = os.path.join(save_dir, f"latent_vae_recon_error_eps_{epsilon}.png")
        plt.savefig(save_path)
        plt.show()  # <-- Show the histogram on screen
        plt.close()
        print(f"[✓] Saved and displayed: {save_path}")


def visualize_magnet_pca_all_eps(magnet, G, save_dir="results"):
    magnet.eval()
    G.eval()
    os.makedirs(save_dir, exist_ok=True)

    for epsilon in config.epsilon_values:
        print(f"\n[MagNet] PCA visualization for ε = {epsilon}")

        # Generate data
        z = torch.randn(200, config.latent_dim).to(device)
        x = G(z)
        adv_z = attacker.enhanced_adaptive_attack(z, detector, epsilon)
        x_adv = G(adv_z)

        # Reconstruct both
        with torch.no_grad():
            recon_x = magnet(x)
            recon_x_adv = magnet(x_adv)

        # Flatten for PCA
        recon_x_np = recon_x.view(recon_x.size(0), -1).detach().cpu().numpy()
        recon_x_adv_np = recon_x_adv.view(recon_x_adv.size(0), -1).detach().cpu().numpy()

        # Combine
        all_recons = np.concatenate([recon_x_np, recon_x_adv_np])
        labels = np.array([0] * recon_x_np.shape[0] + [1] * recon_x_adv_np.shape[0])

        # PCA projection
        pca = PCA(n_components=2)
        z_2d = pca.fit_transform(all_recons)

        # Plot
        plt.figure(figsize=(6, 5))
        plt.scatter(z_2d[labels == 0, 0], z_2d[labels == 0, 1], c='blue', label='Normal', alpha=0.6)
        plt.scatter(z_2d[labels == 1, 0], z_2d[labels == 1, 1], c='red', label='Adversarial', alpha=0.6)
        plt.title(f"MagNet PCA Projection (ε = {epsilon})")
        plt.xlabel("PC1")
        plt.ylabel("PC2")
        plt.legend()
        pca_path = os.path.join(save_dir, f"magnet_pca_eps_{epsilon}.png")
        plt.savefig(pca_path)
        plt.show()
        plt.close()
        print(f"[✓] PCA saved: {pca_path}")



def visualize_latent_vae_pca_all_eps(vae, G, save_dir="results"):
    vae.eval()
    G.eval()
    os.makedirs(save_dir, exist_ok=True)

    for epsilon in config.epsilon_values:
        z = torch.randn(200, config.latent_dim).to(device)
        x_real = G(z)
        adv_z = attacker.enhanced_adaptive_attack(z, detector, epsilon)

        with torch.no_grad():
            recon_normal, _, _ = vae(z)
            recon_adv, _, _ = vae(adv_z)

        z_all = torch.cat([z, adv_z])
        recon_all = torch.cat([recon_normal, recon_adv])
        labels = np.array([0] * len(z) + [1] * len(adv_z))

        z_all_np = recon_all.cpu().numpy()
        pca = PCA(n_components=2)
        z_2d = pca.fit_transform(z_all_np)

        plt.figure(figsize=(8, 6))
        plt.scatter(z_2d[labels == 0][:, 0], z_2d[labels == 0][:, 1], label="Normal", alpha=0.6, c='blue')
        plt.scatter(z_2d[labels == 1][:, 0], z_2d[labels == 1][:, 1], label="Adversarial", alpha=0.6, c='red')
        plt.title(f"LatentVAE PCA Projection (ε={epsilon})")
        plt.xlabel("PCA-1")
        plt.ylabel("PCA-2")
        plt.legend()
        plt.tight_layout()

        save_path = os.path.join(save_dir, f"latent_vae_pca_eps_{epsilon}.png")
        plt.savefig(save_path)
        plt.show()
        plt.close()
        print(f"[✓] PCA saved and displayed: {save_path}")

    

def visualize_latent_explainer_pca_all_eps(explainer, G, save_dir="results"):
    explainer.eval()
    G.eval()
    os.makedirs(save_dir, exist_ok=True)

    for epsilon in config.epsilon_values:
        z = torch.randn(200, config.latent_dim).to(device)
        x_real = G(z)
        adv_z = attacker.enhanced_adaptive_attack(z, detector, epsilon)  # ε looped

        with torch.no_grad():
            recon_normal = explainer(z)
            recon_adv = explainer(adv_z)

        z_all = torch.cat([z, adv_z])
        recon_all = torch.cat([recon_normal, recon_adv])
        labels = np.array([0] * len(z) + [1] * len(adv_z))  # 0 = normal, 1 = adv

        z_all_np = recon_all.cpu().numpy()
        pca = PCA(n_components=2)
        z_2d = pca.fit_transform(z_all_np)

        plt.figure(figsize=(8, 6))
        plt.scatter(z_2d[labels == 0][:, 0], z_2d[labels == 0][:, 1], label="Normal", alpha=0.6, c='blue')
        plt.scatter(z_2d[labels == 1][:, 0], z_2d[labels == 1][:, 1], label="Adversarial", alpha=0.6, c='red')
        plt.title(f"LatentExplainer PCA Projection (ε={epsilon})")
        plt.xlabel("PCA-1")
        plt.ylabel("PCA-2")
        plt.legend()
        plt.tight_layout()

        save_path = os.path.join(save_dir, f"latent_explainer_pca_eps_{epsilon}.png")
        plt.savefig(save_path)
        plt.show()
        plt.close()
        print(f"[✓] PCA saved and displayed: {save_path}")

def visualize_attacks(epsilon=0.1):
    G.eval()
    z = torch.randn(1, config.latent_dim).to(device)
    real_batch = next(iter(train_loader))
    real_img = real_batch[0]
    attacks = {
        "Real": "real",
        "Original": None,
        "FGSM": attacker.fgsm,
        "PGD": attacker.pgd,
        "CW": attacker.cw,
        "Adaptive": lambda z: attacker.enhanced_adaptive_attack(z, detector, epsilon=epsilon)
    }
    num_plots = len(attacks)
    fig, axs = plt.subplots(1, num_plots, figsize=(4*num_plots, 4))
    
    for idx, (name, attack_fn) in enumerate(attacks.items()):
        if name == "Real":
            img = real_img.cpu().detach().squeeze().numpy()
            img = np.transpose(img, (1, 2, 0)) * 0.5 + 0.5
            title = f"{name}\n(CIFAR-10)"
        elif attack_fn is None:
            img = G(z)
            img = img.cpu().detach().squeeze().numpy()
            img = np.transpose(img, (1, 2, 0)) * 0.5 + 0.5
            title = f"{name}\n(Direct Generation)"
        else:
            adv_z = attack_fn(z)
            img = G(adv_z)
            img = img.cpu().detach().squeeze().numpy()
            img = np.transpose(img, (1, 2, 0)) * 0.5 + 0.5
            title = f"{name}\n(ε={epsilon})"
        
        axs[idx].imshow(img)
        axs[idx].set_title(title)
        axs[idx].axis('off')
    
    plt.tight_layout()
    plt.savefig(os.path.join("results", "attack_visualizations.png"))
    plt.show()
    G.train()

def robustness_analysis_adaptive(epsilon=0.1):
    print("Evaluating on standard attacks:")
    metrics_standard = evaluate_detector(attack_type='all', epsilon=epsilon, adaptive=False)
    print("Evaluating on adaptive attack:")
    metrics_adaptive = evaluate_detector(attack_type='adaptive', epsilon=epsilon, adaptive=True)
    return metrics_standard, metrics_adaptive

def formal_stat_tests(xid_metrics, baseline_metrics):
    print("t-test for AUROC:")
    tstat, pval = ttest_ind(xid_metrics[:,0], baseline_metrics[:,0])
    print(f"t-stat = {tstat:.4f}, p-value = {pval:.4f}")
    print("Mann-Whitney U test for AUROC:")
    ustat, upval = mannwhitneyu(xid_metrics[:,0], baseline_metrics[:,0])
    print(f"U-stat = {ustat:.4f}, p-value = {upval:.4f}")

def defense_gan_purify(adversarial_img, G, iterations=100, lr=0.01):
    z = torch.randn(1, config.latent_dim, device=device).requires_grad_(True)
    optimizer = optim.Adam([z], lr=lr)
    target = adversarial_img.detach()
    for i in range(iterations):
        optimizer.zero_grad()
        purified = G(z)
        loss = nn.MSELoss()(purified, target)
        loss.backward()
        optimizer.step()
    return G(z).detach()

def show_tensor_image(img_tensor, title=""):
    img = img_tensor.cpu().detach()
    if img.dim() == 4:
        img = img[0]
    img = img.permute(1, 2, 0).numpy() * 0.5 + 0.5
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")

# Main Pipeline
print("Initializing and training WGAN-GP...")


G = StyleGAN2Generator(config.latent_dim).to(device)
C = StyleGAN2Discriminator().to(device)
train_stylegan2()



print("Initializing detectors...")
torch.cuda.empty_cache()
detector = EnhancedXAIDetector(config.latent_dim).to(device)
baseline_detector = BaselineDetector(config.latent_dim).to(device)
magnet_detector = MagNetDetector().to(device)
latent_explainer = LatentExplainer(config.latent_dim).to(device)
attacker = WGANAttacker(G, C, detector=detector)
# train_latent_explainer(latent_explainer, attacker, G, detector=detector, epochs=20)

latent_vae = LatentVAE(config.latent_dim).to(device)

results_xid = []
results_baseline = []
for eps in config.epsilon_values:
    print(f"\n{'='*40}\nEnhancedXAIDetector on ε={eps}\n{'='*40}")
    train_robust_xai_detector(detector, G)
    metrics = evaluate_detector(attack_type='all', epsilon=eps)
    results_xid.append(metrics)
    
    print(f"\n{'='*40}\nBaseline Detector on ε={eps}\n{'='*40}")
    train_baseline_detector(attack_type='all', epsilon=1)
    b_metrics = evaluate_baseline_detector(attack_type='all', epsilon=eps)
    results_baseline.append(b_metrics)

results_xid = np.array(results_xid)
results_baseline = np.array(results_baseline)
print("\nComparison (EnhancedXAIDetector vs. Baseline Detector):")
print("EnhancedXAIDetector: AUROC = {:.4f} ± {:.4f}, Accuracy = {:.4f} ± {:.4f}, FID = {:.4f} ± {:.4f}".format(
    results_xid[:,0].mean(), results_xid[:,0].std(),
    results_xid[:,1].mean(), results_xid[:,1].std(),
    results_xid[:,2].mean(), results_xid[:,2].std()
))
print("Baseline Detector: AUROC = {:.4f} ± {:.4f}, Accuracy = {:.4f} ± {:.4f}".format(
    results_baseline[:,0].mean(), results_baseline[:,0].std(),
    results_baseline[:,1].mean(), results_baseline[:,1].std()
))
formal_stat_tests(results_xid, results_baseline)
robustness_analysis_adaptive(epsilon=0.1)
visualize_latent_space()
visualize_attacks(epsilon=0.1)

print("\nTraining MagNet Detector...")
train_magnet_detector()
print("\nVisualizing MagNet Detector...")
visualize_magnet_detector(magnet_detector)

evaluate_magnet_detector_all_eps(magnet_detector, G)
visualize_magnet_pca_all_eps(magnet_detector, G)

real_batch = next(iter(train_loader))
imgs = real_batch.to(device)
with torch.no_grad():
    rec_err = magnet_detector.reconstruction_error(imgs)
print("Average reconstruction error (MagNet) on a batch:", rec_err.mean().item())

z = torch.randn(1, config.latent_dim).to(device)
G.eval()
real_img = G(z)
z_adv_example = attacker.enhanced_adaptive_attack(z, detector, epsilon=0.1)
adv_img = G(z_adv_example)
purified_img = defense_gan_purify(adv_img, G, iterations=50, lr=0.01)
plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
show_tensor_image(adv_img, title="Adversarial Image")
plt.subplot(1, 3, 2)
show_tensor_image(purified_img, title="Purified (Defense-GAN)")
plt.subplot(1, 3, 3)
show_tensor_image(real_img, title="Real Image")
plt.tight_layout()
plt.savefig(os.path.join("results", "defense_gan_comparison.png"))
plt.show()

print("\nTraining LatentExplainer...")
train_latent_explainer(latent_explainer, attacker, G, detector=detector, epochs=10)
print("\nVisualizing LatentExplainer...")
# visualize_latent_explainer(latent_explainer)


visualize_latent_explainer_pca_all_eps(latent_explainer, G)
visualize_latent_vae_pca_all_eps(latent_vae, G)

print("\nTraining LatentVAE...")
train_latent_vae(latent_vae, epochs=10)
print("\nVisualizing LatentVAE...")
# visualize_latent_vae(latent_vae, G)
visualize_latent_explainer_histograms_all_eps(latent_explainer)
visualize_latent_vae_histograms_all_eps(latent_vae)



xid_aurocs = evaluate_latent_explainer_roc_all_eps(latent_explainer)
vae_aurocs = evaluate_latent_vae_roc_all_eps(latent_vae)
plot_auroc_vs_epsilon(xid_aurocs, vae_aurocs)



# ======================================================
# === XAI ADD-ONS (does not change training/results) ===
# ======================================================
import matplotlib.pyplot as plt
import numpy as np
import torch

def compute_score_breakdown(z_batch, lambda_entropy=0.1, lambda_one_minus_critic=0.5):
    detector.eval(); latent_explainer.eval(); G.eval(); C.eval()
    z = z_batch.to(device)

    # recon error
    z_hat = latent_explainer(z)
    recon_err = torch.norm(z - z_hat, dim=1)

    # entropy of detector
    p = detector(z).view(-1).clamp(1e-6, 1-1e-6)
    entropy = -(p*torch.log(p) + (1-p)*torch.log(1-p))

    # critic confidence (normalize batch)
    x = G(z)
    c = C(x).view(-1)
    c = (c - c.min()) / (c.max() - c.min() + 1e-8)
    one_minus_critic = 1 - c

    total = recon_err + lambda_entropy*entropy + lambda_one_minus_critic*one_minus_critic
    return {
        "ReconErr": recon_err.detach().cpu().numpy(),
        "Entropy": entropy.detach().cpu().numpy(),
        "OneMinusCritic": one_minus_critic.detach().cpu().numpy(),
        "TotalScore": total.detach().cpu().numpy()
    }

def compute_score_tensor(z_batch, lambda_entropy=0.1, lambda_one_minus_critic=0.5):
    """Same as compute_score_breakdown but keeps torch tensors (needed for saliency)."""
    detector.eval(); latent_explainer.eval(); G.eval(); C.eval()
    z = z_batch.to(device)

    z_hat = latent_explainer(z)
    recon_err = torch.norm(z - z_hat, dim=1)

    p = detector(z).view(-1).clamp(1e-6, 1-1e-6)
    entropy = -(p*torch.log(p) + (1-p)*torch.log(1-p))

    x = G(z)
    c = C(x).view(-1)
    c = (c - c.min()) / (c.max() - c.min() + 1e-8)
    one_minus_critic = 1 - c

    total = recon_err + lambda_entropy*entropy + lambda_one_minus_critic*one_minus_critic
    return total



# ---------- 2) LATENT SALIENCY ----------
def latent_saliency(z_single, lambda_entropy=0.1, lambda_one_minus_critic=0.5):
    z1 = z_single.view(1,-1).to(device).detach().clone().requires_grad_(True)
    total = compute_score_tensor(z1, lambda_entropy=lambda_entropy,
                                 lambda_one_minus_critic=lambda_one_minus_critic).sum()
    total.backward()
    grad = z1.grad.abs().view(-1).cpu().numpy()
    return grad

# ---------- 3) ATTENTION HEATMAPS ----------
def get_attention_weights(detector, z_batch, per_head=False):
    """
    Returns:
      per_head=False: (N, 4, 4) head-averaged attention maps
      per_head=True : (N, H, 4, 4) all heads
    """
    detector.eval()
    z_batch = z_batch.to(device)
    feats = [ext(z_batch) for ext in detector.feature_extractors]     # (N,64) ×4
    X = torch.stack(feats, dim=1)                                     # (N,4,64)

    mha = detector.cross_attention
    if getattr(mha, "batch_first", False):
        q = k = v = X                                                 # (N,4,64)
    else:
        q = k = v = X.transpose(0,1)                                  # (4,N,64)

    _, attn_w = mha(q, k, v, need_weights=True, average_attn_weights=False)
    if isinstance(attn_w, (list, tuple)):
        attn_w = attn_w[0]                                            # safety

    # shapes: (N,H,4,4) or (H,N,4,4) -> make (N,H,4,4)
    if attn_w.dim() == 4 and attn_w.shape[0] == mha.num_heads and not getattr(mha, "batch_first", False):
        attn_w = attn_w.permute(1,0,2,3)

    if per_head:
        return attn_w.detach().cpu().numpy()
    else:
        return attn_w.mean(dim=1).detach().cpu().numpy()

def _plot_attn_auto(A, ax, title="Attention Heatmap (4×4)"):
    """Auto scales vmin/vmax to reveal small differences."""
    import numpy as np
    vmin, vmax = float(A.min()), float(A.max())
    im = ax.imshow(A, cmap="viridis", vmin=vmin, vmax=vmax)
    ax.set_title(title)
    ax.set_xticks(range(4)); ax.set_yticks(range(4))
    return im, vmin, vmax

# ---------- 4) PER-SAMPLE REPORT ----------
def make_xai_report(z_clean, epsilon=0.3, sample_id="demo",
                    lambda_entropy=0.1, lambda_one_minus_critic=0.5,
                    save_scores_csv=True, k_top_sal=20):
    """
    Builds a 2x2 figure: clean vs adv image, saliency bars, attention heatmap.
    Also prints & (optionally) saves the numeric score breakdowns.
    """
    detector.eval(); latent_explainer.eval(); G.eval(); C.eval()

    # adversarial latent
    z_clean = z_clean.view(1,-1).to(device)
    z_adv = attacker.enhanced_adaptive_attack(z_clean, detector, epsilon)

    # images
    with torch.no_grad():
        x_clean = G(z_clean)
        x_adv   = G(z_adv)

    # scores
    sc_clean = compute_score_breakdown(z_clean, lambda_entropy, lambda_one_minus_critic)
    sc_adv   = compute_score_breakdown(z_adv,   lambda_entropy, lambda_one_minus_critic)

    # console print (easy copy to thesis tables)
    def _v(d,k): 
        v = d[k]; return float(v[0]) if isinstance(v, np.ndarray) else float(v)
    print(f"[XAI Scores] sample={sample_id}, eps={epsilon}")
    print("Clean :", {k: round(_v(sc_clean,k), 6) for k in ["ReconErr","Entropy","OneMinusCritic","TotalScore"]})
    print("Adv   :", {k: round(_v(sc_adv,  k), 6) for k in ["ReconErr","Entropy","OneMinusCritic","TotalScore"]})

    # (optional) save a CSV row
    if save_scores_csv:
        csv_path = f"{XAI_ROOT}/scores/score_rows.csv"
        newfile = not os.path.exists(csv_path)
        with open(csv_path, "a", newline="") as f:
            w = csv.writer(f)
            if newfile:
                w.writerow(["sample_id","epsilon","ReconErr_clean","Entropy_clean","OneMinusCritic_clean","Total_clean",
                            "ReconErr_adv","Entropy_adv","OneMinusCritic_adv","Total_adv"])
            w.writerow([
                sample_id, epsilon,
                _v(sc_clean,"ReconErr"), _v(sc_clean,"Entropy"), _v(sc_clean,"OneMinusCritic"), _v(sc_clean,"TotalScore"),
                _v(sc_adv,"ReconErr"),   _v(sc_adv,"Entropy"),   _v(sc_adv,"OneMinusCritic"),   _v(sc_adv,"TotalScore")
            ])

    # saliency on adversarial latent
    sal = latent_saliency(z_adv[0].detach().cpu(),
                          lambda_entropy=lambda_entropy,
                          lambda_one_minus_critic=lambda_one_minus_critic)
    top_idx = sal.argsort()[-k_top_sal:][::-1]

    # attention (auto-scale; head-avg)
    attn = get_attention_weights(detector, z_adv, per_head=False)[0]     # (4,4)

    # ---- plot ----
    fig, axs = plt.subplots(1, 2, figsize=(12,4))

    # (left) saliency bars
    axs[0].bar(range(k_top_sal), sal[top_idx])
    axs[0].set_title("Top-20 Salient Latent Dims")
    axs[0].set_xticks(range(k_top_sal))
    axs[0].set_xticklabels([str(i) for i in top_idx], rotation=60)

    # (right) attention heatmap (auto vmin/vmax)
    im, vmin, vmax = _plot_attn_auto(attn, axs[1], "Attention Heatmap (4×4)")
    plt.colorbar(im, ax=axs[1], fraction=0.046, pad=0.04)
    print(f"[ATTN] head-avg stats -> min={vmin:.6f}, max={vmax:.6f}, range={vmax-vmin:.6f}")

    # Save + show
    out_path = f"{XAI_ROOT}/xai_report_{sample_id}_eps{epsilon}.png"
    plt.tight_layout(); plt.savefig(out_path, dpi=300); plt.show()
    print("✅ Saved XAI report:", out_path)

    # also save the two images separately (useful for thesis panels)
    def _save_img(t, name):
        t = t.detach().cpu()
        if t.shape[1] == 1:
            arr = (t[0,0].numpy()*0.5+0.5)
            plt.imshow(arr, cmap="gray"); plt.axis("off")
        else:
            arr = np.transpose((t[0].numpy()*0.5+0.5), (1,2,0))
            plt.imshow(arr); plt.axis("off")
        p = f"{XAI_ROOT}/{name}"
        plt.savefig(p, dpi=300, bbox_inches="tight", pad_inches=0); plt.close()
        return p
    p_clean = _save_img(x_clean, f"img_clean_{sample_id}_eps{epsilon}.png")
    p_adv   = _save_img(x_adv,   f"img_adv_{sample_id}_eps{epsilon}.png")
    print("🖼️ Saved component images:", p_clean, "and", p_adv)


z = torch.randn(config.latent_dim)  # one random latent
make_xai_report(z, epsilon=0.1, sample_id="test for 0.1")

z = torch.randn(config.latent_dim)  # one random latent
make_xai_report(z, epsilon=0.3, sample_id="test for 0.3")

z = torch.randn(config.latent_dim)  # one random latent
make_xai_report(z, epsilon=0.5, sample_id="test for 0.5")

z = torch.randn(config.latent_dim)  # one random latent
make_xai_report(z, epsilon=0.7, sample_id="test for 0.7")

z = torch.randn(config.latent_dim)  # one random latent
make_xai_report(z, epsilon=0.9, sample_id="test for 0.9")





```
